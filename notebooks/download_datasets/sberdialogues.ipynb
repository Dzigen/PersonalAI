```python
# === qa part ===
# количество qa-пар
# распределение вопросов по категориям
# Сколько диалоговых сессий приходится на одну qa-пару
# Длина вопроса
# Длина ответа
# === conversation ===
# количество сессий
# количество turn-ов в одной сессии
# Длина user-сообщений
# Длина assistant-сообщений
```

```python
evidences_perqa_count = []
assistant_message_len, user_message_len = [], []
for conversations_group in original_dataset:
    for qa_info in conversations_group['qa']:
        evidences_perqa_count.append(len(qa_info['evidence']))
    
    raw_sessions = conversations_group['conversation']
    formated_sessions = []
    counter, flag = 1, True
    while flag:
        session_name = f"session_{counter}"
        cur_session = raw_sessions.get(session_name, None)
        if cur_session is None:
            flag = False
        else:
            formated_sessions.append(cur_session)
            counter += 1
        
    for session in formated_sessions:
        for message_info in session:
            if message_info['speaker'] == 'user':
                user_message_len.append(len(message_info['text']))
            elif message_info['speaker'] == 'assistant':
                assistant_message_len.append(len(message_info['text']))
            else:
                raise ValueError
```

```python
def compute_stats(numbers: List[int]) -> Dict[str, float]:
    return {'median': round(np.median(numbers),5),
            'mean': round(np.mean(numbers),5),
            'std': round(np.std(numbers),5),
            'min': round(min(numbers),5),
            'max': round(max(numbers),5),
            'count': len(numbers)}
```

```python
compute_stats(assistant_message_len)
```

```python
plt.boxplot(assistant_message_len)
plt.grid()
plt.show()
```

```python
compute_stats(user_message_len)
```
