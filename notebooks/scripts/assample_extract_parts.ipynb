```python
import joblib
from tqdm import tqdm
import os

PARTS_DIR = './extract_parts'
```

```python
parts_file_names = os.listdir(PARTS_DIR)
print(len(parts_file_names))
```

```python
extracted_triplets = []
```

```python
for filename in tqdm(parts_file_names):
    cur_name = f"{PARTS_DIR}/{filename}"
    cur_triplets = joblib.load(filename)
    extracted_triplets += cur_triplets

print(len(extracted_triplets))
```
