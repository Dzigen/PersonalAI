```python
def formatted_agenttask_execution_and_print(task_solver: AgentTaskSolver, query: str) -> str:
    print("Query: ", query)
    output = task_solver.solve(query=query)
    print("Result: ", output[0])
    print("="*10)
    return output[0]

def evaluate_manually_agentask(task_solver: AgentTaskSolver, queries_map: Dict[str,List[str]]) -> List[str]:
    results: Dict[str,List[str]] = defaultdict(list)
    for kw, queries in queries_map.items():
        print(f"{kw} "+"="*10)
        for query in queries:
            result = formatted_agenttask_execution_and_print(task_solver, query)
            results[kw].append(result)
    return results
```
