```python
import aerospike
```

```python
config = {
  'hosts': [ ('localhost', 3000) ]
}
```

```python
client = aerospike.client(config).connect()
```

```python
key = ('test', 'demo', 'foo')
print('Working with record key ', key)
```

```python
client.put(key, {
    'name': 'John Doe',
    'age': 32,
    'greeting': 'Hello, World!'
  })
```

```python
(key, metadata, record) = client.get(key)
```

```python
client.close()
```
