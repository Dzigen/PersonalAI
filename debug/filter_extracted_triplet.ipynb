```python
HOTPOTQA_PATHS = {
    'full_triplets': '/home/dzigen/Desktop/PersonalAI/Personal-AI/data/knowledge_graphs/extracted_triplets/hotpotqa_extracted_triplets_full',
    'filtered_triplets': '/home/dzigen/Desktop/PersonalAI/Personal-AI/data/knowledge_graphs/extracted_triplets/hotpotqa_extracted_triplets_filtered',
    'full_contexts': '/home/dzigen/Desktop/PersonalAI/Personal-AI/data/qa_datasets/hotpotqa_distractor_validation/relevant_contexts.csv',
    'filtered_contexts': '/home/dzigen/Desktop/PersonalAI/Personal-AI/data/qa_datasets/v2/hotpotqa_distractor_validation/relevant_contexts.csv'
}

TRIVIAQA_PATHS = {
    'full_triplets': '/home/dzigen/Desktop/PersonalAI/Personal-AI/data/knowledge_graphs/extracted_triplets/triviaqa_extracted_triplets_full',
    'filtered_triplets': '/home/dzigen/Desktop/PersonalAI/Personal-AI/data/knowledge_graphs/extracted_triplets/triviaqa_extracted_triplets_filtered',
    'full_contexts': '/home/dzigen/Desktop/PersonalAI/Personal-AI/data/qa_datasets/trivia_qa_rcwikipedia_validation/relevant_contexts.csv',
    'filtered_contexts': '/home/dzigen/Desktop/PersonalAI/Personal-AI/data/qa_datasets/v2/trivia_qa_rcwikipedia_validation/relevant_contexts.csv'  
}
```

#### hotpot qa

```python
hpqa_full = joblib.load(HOTPOTQA_PATHS['full_triplets'])
```

```python
hpqa_full_df = pd.read_csv(HOTPOTQA_PATHS['full_contexts'])
hpqa_filtered_df = pd.read_csv(HOTPOTQA_PATHS['filtered_contexts'])
```

```python
relevant_indexes = []
for f_cntx in hpqa_filtered_df['context'].to_list():
    finded_index = hpqa_full_df[hpqa_full_df['context'] == f_cntx].index[0]
    relevant_indexes.append(finded_index)
```

```python
hpqa_filtered = [hpqa_full[f_idx] for f_idx in relevant_indexes]
```

```python
joblib.dump(hpqa_filtered, HOTPOTQA_PATHS['filtered_triplets'])
```
